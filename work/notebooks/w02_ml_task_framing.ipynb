## My lane as an ML task (type)

**Task Type:** Scoring

I frame my lane as a **content opportunity scoring** problem. The goal is to assign
an opportunity score to each content item, indicating how valuable it would be to
refresh or update that content. Pages with higher scores should be prioritized for
review and improvement by the content team.

A scoring approach is more useful than a simple classification because it produces
a ranked list of pages instead of only predicting "refresh" or "do not refresh."
This helps the team allocate time and resources to the highest-impact opportunities.

```python
TASK_TYPE = "Scoring"

print(f"ML Task Type: {TASK_TYPE}")
```

## Target or proxy

**Target:** Content Opportunity Score

The model predicts a score that represents how valuable it is to refresh a content page. Since this score is not directly available, it is created as a proxy using historical content performance.

```python
TARGET = "Content Opportunity Score"

print(TARGET)
```

## Success metric

**Metric:** Precision@K

Precision@K measures how many of the top K pages recommended by the model are truly good candidates for refresh. A higher Precision@K means the model is prioritizing the right content.

```python
SUCCESS_METRIC = "Precision@K"

print(SUCCESS_METRIC)
```
